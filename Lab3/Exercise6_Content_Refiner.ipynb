# Exercise 6 — Iterative Content Refiner

## Objective
Revise a draft to fix one specific issue across multiple rounds.

## Inputs
- draft_text
- issue_to_fix

## Expected Output
A revised version of the text that addresses the specified issue over at least three rounds.

```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("GROQ_API_KEY not found. Check your .env file.")

client = Groq(api_key=api_key)
MODEL = "openai/gpt-oss-120b"

print("Groq client initialized successfully.")
```

```python
draft_text = """
Our product is good. It helps people do their work.
Many people can use it. Buy it now because it is the best.
"""

print("ORIGINAL DRAFT")
print(draft_text)
```

```python
def refine_content(draft_text, issue_to_fix):
    if not draft_text.strip():
        raise ValueError("Draft text cannot be empty.")

    if not issue_to_fix.strip():
        raise ValueError("Specify an issue to fix.")

    prompt = f"""
You are a professional content editor.

Revise the draft to fix ONLY the specified issue.

Rules:
1. Focus on the requested issue.
2. Preserve the original meaning and factual claims.
3. Do not invent statistics, evidence, features, or promises.
4. Keep improvements unrelated to the requested issue minimal.
5. Return only the revised draft, without explanations.

Issue to fix:
{issue_to_fix}

Current draft:
{draft_text}
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a careful editor. Fix the specified issue "
                    "without inventing facts or changing the core meaning."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2,
        max_completion_tokens=500
    )

    revised_text = response.choices[0].message.content

    if not revised_text or not revised_text.strip():
        raise ValueError("The model returned an empty revision.")

    return revised_text.strip()
```

```python
issue_round_1 = (
    "Fix grammatical problems and unclear wording. "
    "Keep the original meaning and do not add new claims."
)

draft_round_1 = refine_content(
    draft_text,
    issue_round_1
)

print("ROUND 1 — GRAMMAR AND CLARITY")
print(draft_round_1)
```

```python
issue_round_2 = (
    "Make the wording more professional and suitable for a business audience. "
    "Preserve the meaning and do not add unsupported claims."
)

draft_round_2 = refine_content(
    draft_round_1,
    issue_round_2
)

print("ROUND 2 — PROFESSIONAL TONE")
print(draft_round_2)
```

```python
issue_round_3 = (
    "Remove exaggerated or unsupported promotional language. "
    "Keep the message clear, balanced, and factual without adding new claims."
)

draft_round_3 = refine_content(
    draft_round_2,
    issue_round_3
)

print("ROUND 3 — REMOVE EXAGGERATION")
print(draft_round_3)
```

```python
print("=" * 50)
print("ORIGINAL DRAFT")
print("=" * 50)
print(draft_text)

print("\n" + "=" * 50)
print("ROUND 1 — GRAMMAR AND CLARITY")
print("=" * 50)
print(draft_round_1)

print("\n" + "=" * 50)
print("ROUND 2 — PROFESSIONAL TONE")
print("=" * 50)
print(draft_round_2)

print("\n" + "=" * 50)
print("ROUND 3 — REMOVE EXAGGERATION")
print("=" * 50)
print(draft_round_3)
```

```python
revisions = [
    draft_round_1,
    draft_round_2,
    draft_round_3
]

assert len(revisions) >= 3, "At least three revision rounds are required."

assert all(
    isinstance(revision, str) and revision.strip()
    for revision in revisions
), "Every revision must contain non-empty text."

print("Validation passed: all three revision rounds produced text.")
print("Total revision rounds:", len(revisions))
```
