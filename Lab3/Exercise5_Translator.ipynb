# Exercise 5 — Multi-language Translator

## Objective
Create a prompt that translates a sentence into a target language using a specified level of formality.

## Inputs
- sentence
- target_language
- formality

## Expected Output
A translated sentence that matches the requested language and formality level.

```python
sentence = "Could you please help me with this problem?"

target_language = "Hindi"
formality = "formal"

print("Original Sentence:", sentence)
print("Target Language:", target_language)
print("Formality:", formality)
```

```python
prompt_template = """
You are a professional translator.

Translate the given sentence into {target_language}.

Formality Level: {formality}

Rules:
1. Preserve the original meaning.
2. Use natural and grammatically correct language.
3. Match the requested formality level.
4. Do not add explanations or extra information.
5. Return only the translated sentence.

Sentence:
{sentence}

Translation:
"""

print(prompt_template)
```

```python
final_prompt = prompt_template.format(
    sentence=sentence,
    target_language=target_language,
    formality=formality
)

print(final_prompt)
```

```python
formality_levels = ["formal", "informal"]

for level in formality_levels:
    test_prompt = prompt_template.format(
        sentence=sentence,
        target_language=target_language,
        formality=level
    )

    print("Formality:", level)
    print("Prompt Created:", bool(test_prompt))
    print("-" * 40)
```

## Conclusion

A multi-language translation prompt was created using a sentence, target language, and formality level. The prompt instructs the translator to preserve the original meaning, use natural language, and return only the translated sentence.

The prompt template was tested with formal and informal settings.

Limitation: The current implementation creates translation prompts but does not perform translations using a language model.
