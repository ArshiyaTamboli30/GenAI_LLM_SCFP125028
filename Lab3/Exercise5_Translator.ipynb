# Exercise 5 — Multi-language Translator

## Objective
Create a prompt that translates a sentence into a target language using a specified level of formality.

## Inputs
- sentence
- target_language
- formality

## Expected Output
A translated sentence that matches the requested language and formality level.

```python
import os
from dotenv import load_dotenv
from groq import Groq

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("GROQ_API_KEY not found. Check your .env file.")

client = Groq(api_key=api_key)
MODEL = "openai/gpt-oss-120b"

print("Groq client initialized successfully.")
```

```python
sentence = "Could you please help me with this problem?"
target_language = "Hindi"
formality = "Formal"

print("Original sentence:", sentence)
print("Target language:", target_language)
print("Formality:", formality)
```

```python
prompt_template = """
You are a professional translator.

Translate the sentence into {target_language}.

Requirements:
- Formality: {formality}
- Preserve the original meaning.
- Use natural, grammatically correct language.
- Do not add explanations or extra information.
- Return only the translated sentence.

Sentence:
{sentence}
"""

prompt = prompt_template.format(
    target_language=target_language,
    formality=formality,
    sentence=sentence
)

print("Prompt created successfully.")
```

```python
def translate_sentence(sentence, target_language, formality):
    if not sentence.strip():
        raise ValueError("Sentence cannot be empty.")

    if not target_language.strip():
        raise ValueError("Target language cannot be empty.")

    if formality not in {"Formal", "Informal"}:
        raise ValueError("Formality must be Formal or Informal.")

    prompt = prompt_template.format(
        target_language=target_language,
        formality=formality,
        sentence=sentence
    )

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": (
                    "Translate accurately into the requested language "
                    "and formality. Return only the translation."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2,
        max_completion_tokens=300
    )

    translation = response.choices[0].message.content

    if not translation or not translation.strip():
        raise ValueError(
            "The model returned an empty translation. Try again."
        )

    return translation.strip()
```

```python
translation = translate_sentence(
    sentence,
    target_language,
    formality
)

print("Original:", sentence)
print("Translated:", translation)
print("Formality:", formality)
```

```python
informal_translation = translate_sentence(
    sentence="Can you help me?",
    target_language="Hindi",
    formality="Informal"
)

print("Original: Can you help me?")
print("Informal Hindi:", informal_translation)
```

```python
english_to_french = translate_sentence(
    sentence="Thank you for your help.",
    target_language="French",
    formality="Formal"
)

print("Original: Thank you for your help.")
print("Formal French:", english_to_french)
```
